```
# Antes de decidir si voy a trabajar un dataset visualizo para tener una idea de qué me voy a encontrar

import pandas as pd                                           # importo pandas para poder leer el archivo con el que voy a trabajar

df = pd.read_excel("/content/4_classification_diabetes.xlsx") # lee el archivo
df.head()                                                     # muestra las primeras 5 filas de un dataframe (de la 0 a la 4)
df.shape                                                      # muestra las dimensiones de un dataframe
df.info()                                                     # muestra número de filas y columnas, nombre de columnas, número de valores No Null, tipo de datos por columna
```

```
#                                                                       ABSTRACT

# En este proyecto se trabaja con el dataset de Clasificación de Diabetes, el cual contiene información relacionada con el estado de salud, los hábitos de vida y
# las condiciones socioeconómicas de personas adultas. El conjunto de datos incluye variables asociadas al diagnóstico de diabetes, presión arterial alta,
# colesterol elevado, índice de masa corporal, consumo de tabaco y alcohol, actividad física, hábitos alimenticios y antecedentes de enfermedades cardiovasculares.
# Además, se incorporan variables vinculadas al acceso al sistema de salud, la percepción del estado general de salud, la salud mental y física, el sexo, el grupo
# etario, el nivel educativo y el nivel de ingresos del hogar.

# La diabetes es una enfermedad crónica muy presente en la población a nivel mundial y representa un importante problema de salud pública. Su estudio resulta
# relevante debido a las complicaciones que puede generar y al impacto que tiene en la calidad de vida de las personas. Por este motivo, es fundamental analizar
# los factores que pueden estar relacionados con su aparición y desarrollo.

# El objetivo principal de este trabajo es analizar el dataset con el fin de identificar patrones y relaciones entre el estado de diabetes y distintos factores
# de riesgo. Para ello, se realiza un análisis exploratorio de los datos utilizando visualizaciones y resúmenes estadísticos. Se emplean gráficos univariados para
# analizar la distribución de las variables, gráficos bivariados para estudiar la relación entre pares de variables y gráficos multivariados para evaluar la
# interacción entre tres o más variables de manera simultánea.

# Asimismo, se identifican los valores perdidos presentes en el conjunto de datos y se analiza su posible impacto en los resultados. Este análisis exploratorio
# permite obtener una primera comprensión de los datos y sienta las bases para futuros estudios más avanzados, como la construcción de modelos de clasificación
# o predicción relacionados con la diabetes.

# Hipótesis principales:

# ¿Está relacionada la diabetes con el Indice de Masa Corporal (BMI)?
# ¿Está relacionada la diabetes con la edad del individuo?
# ¿Está relacionada la diabetes con la cantidad de días de actividad física por mes?
# ¿Las variables que más se relacionan con la diabetes están relacionadas entre sí?


# ----------------------------------------------------------------------------------------------------------------------------------------------------------------#

# GRAFICO UNIVARIADO (DE 1 VARIABLE)            DISTRIBUCION (INDICE) DE MASA CORPORAL (BMI)

# La finalidad de observar esta distribución es averiguar si en la población de esta muestra predomina la obesidad, sabiendo que para que una persona sea
# considerada obesa, la misma deberá presentar valores de BMI iguales o superiores a 30.
# Para ello se realiza un histograma de BMI



import matplotlib.pyplot as plt                             # se importa la librería marplotlib para realizar el gráfico
import seaborn as sns                                       # se importa la librería seaborn para reealizar el histograma
import numpy as np                                          # se importa la librería numpy para definir los intervalos que yo quiera

bins = np.arange(10,100,5)                                  # defino los intervalos en el arange de 10 a 100 en intervalos de 5

plt.figure(figsize=(8, 5))
sns.histplot(df['BMI'], bins=bins, kde=True)                # realiza el histograma de BMI, lo divide en bins definidos arriba, kde muestra la curva de densidad
plt.xlabel("Índice de Masa Corporal (BMI)")                 # nombre eje x
plt.ylabel("Frecuencia")                                    # nombre eje y
plt.title("Distribución del Índice de Masa Corporal (BMI)") # título del gráfico
plt.show()                                                  # hace un display del gráfico

# Le pedimos los valores estadísticos clásicos
media_bmi = df['BMI'].mean()                                # promedio o media (valor medio de la muestra)
mediana_bmi = df['BMI'].median()                            # mediana (valor que divide el 50 % de los datos)
moda_bmi = df['BMI'].mode()                                 # moda (valor más repetido)
desvio_bmi = df['BMI'].std()                                # desvío estandar
asimetria_bmi = df['BMI'].skew()                            # asimetría

# se presentan los valores de las variables definidas arriba
print(f"Media = {media_bmi}")
print(f"Mediana = {mediana_bmi}")
print(f"Moda = {moda_bmi}")
print(f"Desvío estandar = {desvio_bmi}")
print(f"Asimetría = {asimetria_bmi}")


# Se obtienen promedio de BMI de 28.4 y mediana de 27. Esto indica que el promedio de la muestra se encuentra cerca de los valores de obesidad y que la mitad
# de la misma está por debajo del límite de obesidad. Es decir, al menos el 50 % no es obeso
# De nuestro gráfico de BMI podemos observar a simple vista en la curva de densidad,que la moda de nuestra muestra se encuentra entre los 25 y 30 BMI. Es decir
# que es inferior pero cercana al límite de obesidad. En los datos obtenemos un valor de moda de 27, lo que concuerda con el gráfico
# Obtenemos un desvío estádar de 6.6 que nos indica que tan agrupados están respecto de la media
# Se obtiene un valor de asimtría positivo que se observa en el gráfico como la 'cola' a la derecha
```

```
#                                                       PORCENTAJE DE DIABETES

# Quiero averiguar con qué porcentajes de diabetes, prediabetes y sin diabetes estoy trabajando

conteo_diabetes = df['Diabetes_012'].value_counts().sort_index()          # cuenta la cantidad de casos 0, 1 y 2 de la columna Diabetes_012 y las ordena

labels = ['No diabetes', 'Prediabetes', 'Diabetes']                       # asigna las etiquetas (en ese orden porque están ordenadas 0, 1 y 2)

plt.figure(figsize=(6, 6))
plt.pie(conteo_diabetes, labels=labels, autopct='%1.1f%%', startangle=90) # grafico tipo torta, con porcentajes correspondientes y etiquetas
# el 'startangle=90' hace que arranque a graficar en 90 grados, PI/2

plt.title('Distribución del estado de diabetes en la población')          # título del gráfico
plt.show()                                                                # ace un display del gráfico

# Obtengo en mi muestra un 84% de individuos sin diabetes, un 1.8% con prediabetes y un 13.9% con diabetes
```

```
# ----------------------------------------------------------------------------------------------------------------------------------------------------------------#
# GRAFICO BIVARIADO (DE 2 VARIABLES)            INDICE DE MASA CORPORAL SEGUN EL ESTADO DE LA DIABETES

# La finalidad de este gráfico es observar la distribución de BMI en una población con diabetes (2), prediabetes(1) y sin diabetes(0)
# Para ello se realiza un gráfico tipo boxplot, donde se podrán detectar valores atípicos

plt.figure(figsize=(8, 5))
sns.boxplot(data=df,x="Diabetes_012",y="BMI")                                             # grafica el boxplot de diabetes en el eje x y BMI en el eje y

plt.axhline(y=30,color='red', linestyle='--', linewidth=1, label='BMI = 30 (Obesidad)')   # quiero que marque el límite entre obesidad y no obesidad (y=30)

plt.xlabel("Estado de Diabetes (0: No, 1: Prediabetes, 2: Diabetes)")                     # nombre eje x
plt.ylabel("Índice de Masa Corporal (BMI)")                                               # nombre eje x
plt.title("Distribución del BMI según estado de diabetes")                                # título del gráfico
plt.legend()                                                                              # hace un display de la leyenda (etiqueta de BMI = 30 (Obesidad))
plt.show()                                                                                # hace un display del gráfico

# Se observa en el gráfico promedios de BMI inferiores en el grupo sin diabetes respecto a los grupos de diabetes y pre diabetes.
# El cuartil (Q3) de BMI del grupo sin diabetes es igual o inferior a las medias (promedios) del los demás grupos. Sin embargo, todos los grupos presentan
# valores de BMI elevados, lo que indica obesidad muy marcada indistintamente de si el individuo presenta o no algún estado de diabetes
```

```
#                                               DIABETES SEGUN EL GRUPO ETARIO

# La finalidad de este gráfico es observar la distribución etaria en una población con diabetes, prediabetes y sin diabetes

# Se realizará un grafico de barras ...agregar mas texto

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x="Age", hue="Diabetes_012")                 # realiza un gráfico de barras dividiendo por tipo de diabetes

plt.xlabel("Grupo etario")                                          # nombre eje x
plt.ylabel("Cantidad de personas")                                  # nombre eje y
plt.title("Estado de diabetes según grupo etario")                  # título del gráfico
plt.legend(title="Diabetes (0: No, 1: Prediabetes, 2: Diabetes)")   # hace un display de la leyenda
plt.show()                                                          # hace un display del gráfico

# Se observa que el promedio de edad del grupo sin diabetes es menor al promedio de edad del grupo con diabetes.
```

```
#                                               DIABETES SEGUN ACTIVIDAD FíSICA

# La finalidad de este gráfico es observar la distribución etaria en una población con diabetes, prediabetes y sin diabetes

# Se realizará un grafico de barras ...agregar mas texto

plt.figure(figsize=(16, 6))                                         # se agranda la dimensión x del gráfico ya que la variable PhysHlth tiene mayor rango
sns.countplot(data=df, x="PhysHlth", hue="Diabetes_012")            # realiza un gráfico de barras dividiendo por tipo de diabetes

plt.xlabel("Actividad física por mes")                              # nombre eje x
plt.ylabel("Cantidad de personas")                                  # nombre eje y
plt.title("Estado de diabetes según actividad física por mes")      # título del gráfico
plt.legend(title="Diabetes (0: No, 1: Prediabetes, 2: Diabetes)")   # hace un display de la leyenda
plt.show()                                                          # hace un display del gráfico

# No se observa que los individuos de la muestra realicen actividad física. No se obtienen conclusiones
```

```
# GRAFICO MULTIVARIADO (DE 3 VARIABLES O MAS)            BMI, GRUPO ETARIO Y ESTADO DE LA DIABETES

# Se realizará un gráfico de dispersión entre el grupo etario y el BMI, con la consideración de los grupos de Diabetes. Y se calculará la regresión
# lineal y el valor R² para observar si existe una correlación entre las variables edad - BMI y edad y Diabetes

from scipy.stats import linregress                                              # importo la librería scipy porque quiero la regresión lineal

resultado = linregress(df['Age'], df['BMI'])                                    # calcula la regresión lineal entre la edad y el BMI
r2 = resultado.rvalue ** 2                                                      # calculo el R²

resultado_B = linregress(df['Age'], df['Diabetes_012'])                         # calcula la regresión lineal entre la edad y Diabetes
r2_B = resultado_B.rvalue ** 2                                                  # calculo el R²

print(f"R² Edad - BMI= {r2:.3f}")                                               # se presenta el coeficiente de correlación entre de Edad - BMI
print(f"R² Edad - Diabetes= {r2_B:.3f}")                                        # se presenta el coeficiente de correlación entre de Edad - Diabetes

plt.figure(figsize=(10, 6))

sns.scatterplot( data=df, x="Age", y="BMI", hue="Diabetes_012", alpha=0.6)      # grafica el gráfico de dispersión de Edad - BMI con los datos de Diabetes

plt.axhline(30, color='red', linestyle='--', label='BMI = 30')                  # límite entre obesidad y no obesidad

plt.xlabel("Grupo etario")                                                      # nombre eje x
plt.ylabel("Índice de Masa Corporal (BMI)")                                     # nombre eje y
plt.title("Relación entre BMI, edad y estado de diabetes")                      # título del gráfico
plt.legend(title="Estado de Diabetes")                                          # hace un display de la leyenda
plt.show()                                                                      # hace un display del gráfico

# En este gráfico se puede observar mayor cantidad de individuos con diabetes y pre diabetes (colores oscuros) por encima de la línea que marca la obesidad.
# Se aprecia distingue también una mayor cantidad de individuos con diabetes y pre diabetes en la mitad superior de los grupos etarios
# Se obtiene un R² Edad - BMI = 0.001. Un valor bajo para una posible correlación entre BMI y grupo etario del individuo.
```
